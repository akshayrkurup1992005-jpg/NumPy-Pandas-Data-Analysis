## 1. Creating a One-Dimensional NumPy Array
Create an array containing the daily temperatures
recorded during the first week.

```python
# Import NumPy
import numpy as np

# Store Week 1 temperature values
temperatures_w1 = np.asarray(
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
)

# Print the temperature values
print("Temperature Data - Week 1:")
print(temperatures_w1)
```

## 2. Checking Array Properties
Find the dimensions, data type and total number
of values in the Week 1 temperature array.

```python
# Check array dimensions
print("Dimensions:", temperatures_w1.ndim)

# Check array shape
print("Array Shape:", temperatures_w1.shape)

# Check data type
print("Element Type:", temperatures_w1.dtype)

# Count total values
print("Total Values:", len(temperatures_w1))
```

## 3. Temperature Calculations
Convert Celsius values into Fahrenheit and calculate
the highest, lowest and average temperatures.

```python
# Convert temperatures into Fahrenheit
fahrenheit = temperatures_w1 * 1.8 + 32

# Calculate temperature statistics
highest = temperatures_w1.max()
lowest = temperatures_w1.min()
average = temperatures_w1.mean()

# Display the results
print("Fahrenheit Values:", np.round(fahrenheit, 2))
print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)
print("Average Temperature:", round(average, 2))
```

## 4. Extracting Array Elements
Retrieve the first three days, weekend values
and middle three days using array slicing.

```python
# Get values from Monday to Wednesday
first_three = temperatures_w1[:3]

# Get Saturday and Sunday values
weekend = temperatures_w1[5:]

# Get the middle three values
middle_three = temperatures_w1[2:5]

# Print extracted values
print("First Three Days:", first_three)
print("Weekend Values:", weekend)
print("Middle Three Days:", middle_three)
```

## 5. Creating a Two-Dimensional Array
Combine the temperature records of Week 1 and Week 2
into a single two-dimensional NumPy array.

```python
# Store Week 2 temperature values
week2 = [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]

# Combine both weeks into a 2D array
temperatures = np.vstack((temperatures_w1, week2))

# Display the combined array
print("Two Weeks Temperature Records:")
print(temperatures)
```

## 6. Exploring the Two-Dimensional Array
Check the array properties and extract weekly
temperature records and weekend values.

```python
# Display array properties
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

# Extract individual weekly records
week_one = temperatures[0, :]
week_two = temperatures[1, :]

# Extract weekend records from both weeks
weekend_values = temperatures[:, 5:]

# Display the results
print("\nWeek 1 Records:", week_one)
print("Week 2 Records:", week_two)
print("Weekend Records:\n", weekend_values)
```

## 1. Creating a Pandas Series
Create a Series containing student marks with
custom rank labels.

```python
# Import Pandas
import pandas as pd

# Define marks and rank labels
mark_values = [95, 92, 89, 85, 80]
rank_labels = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']

# Construct the Series
marks = pd.Series(data=mark_values, index=rank_labels)

# Display the Series
print("Student Marks:")
print(marks)
```

## 2. Accessing and Filtering Series Data
Retrieve student marks using different indexing
methods and filter students scoring above 90.

```python
# Access the first rank using its index position
first_rank = marks[marks.index[0]]
print("First Rank Mark:", first_rank)

# Retrieve the top three ranks using label slicing
top_students = marks.loc['Rank1':'Rank3']
print("\nTop Three Ranks:")
print(top_students)

# Retrieve the third rank using iloc
third_rank = marks.iloc[2]
print("\nThird Rank Mark:", third_rank)

# Select students who scored above 90
high_scores = marks.loc[marks.gt(90)]
print("\nStudents Scoring Above 90:")
print(high_scores)
```

## 3. Modifying Pandas Series
Update the first rank mark, remove the last rank
and calculate CGPA for the remaining students.

```python
# Update the first rank mark
marks.at['Rank1'] = 100

# Display updated marks
print("Updated Student Marks:")
print(marks)

# Remove the last rank
marks.drop(index='Rank5', inplace=True)

print("\nSeries After Removing Last Rank:")
print(marks)

# Calculate CGPA
cgpa = marks.div(10)

print("\nCalculated CGPA:")
print(cgpa)
```

## 1. Creating a Transaction DataFrame
Create a DataFrame using transaction records,
including product category, region and amount.

```python
# Define transaction records
records = [
    (101, 'Electronics', 'North', 200),
    (102, 'Clothing', 'South', 150),
    (103, 'Electronics', 'North', 300),
    (104, 'Furniture', 'East', 450),
    (105, 'Clothing', 'West', 200),
    (106, 'Electronics', 'North', 250),
    (107, 'Furniture', 'East', 300),
    (108, 'Clothing', 'West', 180),
    (109, 'Furniture', 'South', 350),
    (110, 'Electronics', 'North', 400)
]

# Convert records into a DataFrame
transactions = pd.DataFrame(
    records,
    columns=['TransactionID', 'ProductCategory', 'Region', 'Amount']
)

# Display transaction records
print("Transaction Details:")
print(transactions)
```

## 2. Exploring Transaction Data
Inspect the DataFrame, select columns, filter records,
identify unique values and calculate regional averages.

```python
# Display all transaction records
display(transactions)
```

```python
# Retrieve the first five records
print("Beginning Records:")
display(transactions.head(5))

# Retrieve the last five records
print("Ending Records:")
display(transactions.tail(5))
```

```python
# Check the number of rows and columns
print("Dataset Dimensions:", transactions.shape)

# List all column names
print("\nAvailable Columns:")
print(list(transactions))

# Display the data types
print("\nColumn Data Types:")
print(transactions.dtypes)

# Display DataFrame information
print("\nDataset Summary:")
transactions.info()
```

```python
# Select product category and transaction amount
selected_columns = transactions.filter(
    items=['ProductCategory', 'Amount']
)

print("Selected Transaction Details:")
display(selected_columns)
```

```python
# Select the final three columns
last_columns = transactions.iloc[:, -3:]

print("Last Three Columns:")
display(last_columns)
```

```python
# Find North region transactions with amount above 200
north_sales = transactions.query(
    "Region == 'North' and Amount > 200"
)

print("Filtered North Region Transactions:")
display(north_sales)
```

```python
# Count transactions in each product category
category_summary = transactions['ProductCategory'].value_counts()

print("Category-wise Transaction Count:")
print(category_summary)
```

```python
# Extract distinct region names
regions = transactions['Region'].drop_duplicates().tolist()

print("Available Regions:")
print(regions)
```

```python
# Calculate average transaction amount for every region
regional_summary = transactions.groupby(
    'Region',
    as_index=True
).agg(
    Average_Amount=('Amount', 'mean')
)

print("Regional Average Transaction Amount:")
print(regional_summary)
```

## 3. Modifying Transaction Data
Update a transaction amount, create a discount column,
remove a specific transaction and delete the discount column.

```python
# Locate the row belonging to TransactionID 102
row_position = transactions.index[
    transactions['TransactionID'].eq(102)
][0]

# Update its transaction amount
transactions.at[row_position, 'Amount'] = 165

# Display the modified data
print("After Updating Transaction Amount:")
display(transactions)
```

```python
# Calculate a 10% discount on each transaction
transactions['Discount'] = transactions['Amount'].mul(0.10)

# Display the updated DataFrame
print("After Calculating Discounts:")
display(transactions)
```

```python
# Identify the index of TransactionID 109
remove_index = transactions.index[
    transactions['TransactionID'].eq(109)
]

# Remove the identified transaction
transactions.drop(index=remove_index, inplace=True)

# Display the remaining transactions
print("After Removing Transaction 109:")
display(transactions)
```

```python
# Remove the Discount column
del transactions['Discount']

# Display the final DataFrame
print("Final Transaction Dataset:")
display(transactions)
```
